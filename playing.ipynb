```python
from typing import TypedDict, List

class VideoCandidate(TypedDict):
    video_id:     str
    title:        str
    description:  str
    channel_name: str
    
    # view_count:   int
    # duration:     str
    # url:          str
    # has_captions: bool


response = youtube.search().list(
        part="snippet",
        q="whats up",
        type="video",
        maxResults=10,
        order="relevance",
        videoCaption="closedCaption"
    ).execute()

items = response["items"]


ids = []
for item in items:
    ids.append(item['id']['videoId'])

details_response = youtube.videos().list(
    part="snippet,statistics,contentDetails",
    id=ids
).execute()

items_destails = details_response["items"]
items_list : List[VideoCandidate] = []

for item, video in zip(items,items_destails):
    video_id    = item["id"]["videoId"]
    title       = item["snippet"]["title"]
    description = item["snippet"]["description"]
    channel     = item["snippet"]["channelTitle"]

    view_count = video["statistics"]["viewCount"]
    duration   = video["contentDetails"]["duration"]

    items_list.append(
        {
            "video_id" : video_id,
            "title" : title,
            "description" : description,
            "channel_name" : channel
        }
    )
items_list
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")
```

```python
QUERY_FORMULATOR_PROMPT = """
You are a YouTube search expert.
Given a user question, generate exactly 3 short keyword-focused YouTube search queries
that would find videos answering it.

User question: {question}

Rules:
- Each query should be 3 to 6 words
- Focus on keywords, not full sentences
- Each query should approach the question from a different angle
- Return one query per line, nothing else, no numbering, no punctuation
"""
question = "What is love?"
prompt = QUERY_FORMULATOR_PROMPT.format(question=question)
response = llm.invoke(prompt)
queries = response.content.split("\n")

```

```python
RELEVANCE_RANKER_PROMPT = """
You are evaluating whether a YouTube video answers a user question.

User question: {question}

Video title: {title}
Video description: {description}

Score this video from 0.0 to 1.0 based on how well it answers the question.
Return a JSON object with exactly two fields:
- score: a float between 0.0 and 1.0
- explanation: one sentence explaining the score

Return nothing else. The response should be a string starting with {{ and ending with }}. nothing else.
"""
```

```python
question = "what is llm"
title = "I Hacked This Temu Router. What I Found Should Be Illegal."
description = """
Learn how to hack and reverse engineer like this: https://go.lowlevel.tv/StackSmash-RE

🏫 MY COURSES
Sign-up for my FREE 3-Day C Course: https://lowlevel.academy

⌨️ KEYBOARD
Like what you hear? Grab a Q5 at https://go.lowlevel.tv/keyboard

🔥COME HANG OUT   
Check out my other stuff: https://lowlevel.tv
"""
prompt = RELEVANCE_RANKER_PROMPT.format(
            question=question,
            title=title,
            description=description
        )
response = llm.invoke(prompt)
```
